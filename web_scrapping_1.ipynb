Importing the essential libraries

```python
import requests
from bs4 import  BeautifulSoup
```

Checking the response 

```python
url="https://www.oyorooms.com/hotels-in-chennai/?gad_campaignid=1701551904&gad_source=google&gclid=Cj0KCQjwv4XUBhDBARIsAE6bQURyjnNdPXAQkugbgFSbiyqyn-9ObIc0VOFF0JmlsgFXGMXSJp5uKCIaAnzIEALw_wcB&page=2&utm_adgroup=143959616297&utm_campaign=India_SEM_Brand_generic&utm_campaignid=1701551904&utm_content=751298002022&utm_device=c&utm_keyword=oyo&utm_medium=cpc&utm_source=google"
response=requests.get(url,headers={"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3"})
if response.status_code!=200:
    print("Error: Unable to fetch the webpage.")
else:
     print(response.status_code)
```

Converting the data

```python
soup=BeautifulSoup(response.content,"html.parser")
```

Scraping the data

```python
hotels_card = soup.find_all('div',class_="oyo-cell--12-col oyo-cell--8-col-tablet oyo-cell--4-col-phone")
hotels_card
```

Loop through the first 5 pages

```python
all_hotels_data = []
for page_num in range(1, 6):
     page_url = f"https://www.oyorooms.com/hotels-in-chennai/?gad_campaignid=1701551904&gad_source={page_num}&gclid=Cj0KCQjwv4XUBhDBARIsAE6bQURyjnNdPXAQkugbgFSbiyqyn-9ObIc0VOFF0JmlsgFXGMXSJp5uKCIaAnzIEALw_wcB&page=2&utm_adgroup=143959616297&utm_campaign=India_SEM_Brand_generic&utm_campaignid=1701551904&utm_content=751298002022&utm_device=c&utm_keyword=oyo&utm_medium=cpc&utm_source=google"
     print(f"Scraping page {page_num}")
```

```python
for item in hotels_card:
    hotels=soup.find_all('div',class_="oyo-row oyo-row--no-spacing hotelCardListing")
```

Scraping the data

```python
for hotel in hotels:
    title = hotel.h3["title"]
    Address=hotel.find('span',class_="u-line--clamp-2")["title"]
    num_rating=hotel.find('span',class_="hotelRating__ratingSummary hotelRating__rating--clickable")
    star_rating=hotel.find('span',class_="is-fontBold hotelRating__rating hotelRating__rating--verygood hotelRating__rating--clickable")
    price=hotel.find('span',class_="listingPrice__finalPrice listingPrice__finalPrice--black").text
    
    all_hotels_data.append({"title": title,
                "Address":Address,
                "num_rating":num_rating,
                "star_rating":star_rating,
                "price": price,})
```

Converting into a csv file

```python
import pandas as pd
df = pd.DataFrame(all_hotels_data)
df.to_csv("hotels_data.csv", index=False)
print("Saved:", "hotels_data.csv")
```
